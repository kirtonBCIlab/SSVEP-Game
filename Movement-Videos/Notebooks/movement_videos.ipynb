# Import Packages

```python
import pandas as pd
from ast import literal_eval

import os
import sys
sys.path.append("../Functions")

from map_templates import  Map1, Map2
from vis_video import visualize_map_video
from file_info import extract_file_components
```

# Import Data

```python
# Import a single movement tracking csv file
data_path = "../Data/P047/movements_P047_BW_M2.csv"
df = pd.read_csv(data_path)

# Extract information from path name
file_components = extract_file_components(data_path)

# Set variables
participant_id = file_components['participant']
stim_name = file_components['stim_name']
map_number = file_components['map']
filename = file_components['file']

# Extract relevant data from movement data sheet
coords = df[['Old Position', 'New Position', 'Intended Movement Direction', "Keypress Used"]]
```

# Flip coordinates to be compatible with map code 

```python
# Initalize empty list to hold flipped coords
flipped_coords = []
for index, row in coords.iterrows():
    old_pos = row['Old Position']
    new_pos = row['New Position']
    intended_dir = row['Intended Movement Direction']
    keypress = row['Keypress Used']

    # Flip coords for Old and New Position
    old_x, old_y = [p.strip() for p in old_pos.strip("()").split(",")]
    new_x, new_y = [p.strip() for p in new_pos.strip("()").split(",")]

    # Handle intended_dir with try-except
    try:
        if pd.isna(intended_dir) or str(intended_dir).strip() == "":
            flipped_int_dir = "()"
        else:
            int_dir_x, int_dir_y = [p.strip() for p in str(intended_dir).strip("()").split(",")]
            flipped_int_dir = f"({int(int_dir_y)},{int(int_dir_x)})"
    except (ValueError, AttributeError):
        # If parsing fails, set to empty
        flipped_int_dir = "()"
    
    # Re-format to save flipped coords
    flipped_old = f"({int(old_y)},{int(old_x)})"
    flipped_new = f"({int(new_y)},{int(new_x)})"
    
    # Add flipped coords to the list
    flipped_coords.append((flipped_old, flipped_new, flipped_int_dir, keypress))

# Convert list to data frame
flipped_df = pd.DataFrame(flipped_coords, columns=['Old Position', 'New Position', 'Intended Movement Direction', "Keypress Used"])
```

# Generate video that shows movement through map

```python
# Convert to coordinate tuples
old_positions = [literal_eval(pos) for pos in flipped_df['Old Position']]
new_positions = [literal_eval(pos) for pos in flipped_df['New Position']]
intended_directions = [literal_eval(pos) for pos in flipped_df['Intended Movement Direction']]
keypresses = flipped_df['Keypress Used'].tolist()

# Get the correct map class
m = Map1 if map_number == 1 else Map2

# Create the video visualization
visualize_map_video(
    participant_id,
    stim_name,
    m,
    old_positions,
    new_positions,
    intended_directions,
    keypresses,
    save_video=True,
    video_filename=f"C:\\Users\\admin\\Documents\\Masters\\Path-Generation\\Data\\{participant_id}\\{filename}.mp4"
)
```
