```python
data_files = [
    # Skip P001
    "../Data/P002/movements_P002_BW_M1.csv",
    "../Data/P002/movements_P002_C1S1_M2.csv",
    "../Data/P003/movements_P003_BW_M1.csv",
    "../Data/P003/movements_P003_C4S1_M2.csv",
    "../Data/P004/movements_P004_BW_M2.csv",
    "../Data/P004/movements_P004_C4S1_M1.csv",
    "../Data/P005/movements_P005_BW_M1.csv",
    "../Data/P005/movements_P005_C4S2_M2.csv",
    "../Data/P006/movements_P006_BW_M1.csv",
    "../Data/P006/movements_P006_C4S2_M2.csv",
    "../Data/P007/movements_P007_BW_M1.csv",
    "../Data/P007/movements_P007_C4S1_M2.csv",
    "../Data/P008/movements_P008_BW_M2.csv",
    "../Data/P008/movements_P008_C4S2_M1.csv",
    "../Data/P009/movements_P009_BW_M1.csv",
    "../Data/P009/movements_P009_C4S1_M2.csv",
    "../Data/P010/movements_P010_BW_M2.csv",
    "../Data/P010/movements_P010_C4S1_M1.csv",
    "../Data/P011/movements_P011_C3S1_M1.csv",
    "../Data/P011/movements_P011_BW_M2.csv",
    "../Data/P012/movements_P012_BW_M2.csv",
    "../Data/P012/movements_P012_C3S1_M1.csv",
    "../Data/P013/movements_P013_C4S1_M1.csv",
    "../Data/P013/movements_P013_BW_M2.csv",
    "../Data/P014/movements_P014_BW_M1.csv",
    "../Data/P014/movements_P014_C4S1_M2.csv",
    "../Data/P015/movements_P015_BW_M1.csv",
    "../Data/P015/movements_P015_C4S1_M2.csv",
    # Skip P016
    "../Data/P017/movements_P017_BW_M1.csv",
    "../Data/P017/movements_P017_C4S2_M2.csv",
    "../Data/P018/movements_P018_C4S1_M2.csv",
    "../Data/P018/movements_P018_BW_M1.csv",
    "../Data/P019/movements_P019_BW_M2.csv",
    "../Data/P019/movements_P019_C3S1_M1.csv",
    "../Data/P020/movements_P020_C3S1_M1.csv",
    "../Data/P020/movements_P020_BW_M2.csv",
    "../Data/P021/movements_P021_BW_M1.csv",
    "../Data/P021/movements_P021_C3S3_M2.csv",
    "../Data/P022/movements_P022_C4S2_M1.csv",
    "../Data/P022/movements_P022_BW_M2.csv",
    # Skip P023
    "../Data/P024/movements_P024_C4S1_M1.csv",
    "../Data/P024/movements_P024_BW_M2.csv",
    "../Data/P025/movements_P025_BW_M1.csv",
    "../Data/P025/movements_P025_C2S1_M2.csv",
    "../Data/P026/movements_P026_BW_M2.csv",
    "../Data/P026/movements_P026_C3S2_M1.csv",
    "../Data/P027/movements_P027_BW_M1.csv",
    "../Data/P027/movements_P027_C4S1_M2.csv",
    "../Data/P028/movements_P028_C3S1_M2.csv",
    "../Data/P028/movements_P028_BW_M1.csv",
    "../Data/P029/movements_P029_BW_M2.csv",
    "../Data/P029/movements_P029_C3S1_M1.csv",
    "../Data/P030/movements_P030_BW_M1.csv",
    "../Data/P030/movements_P030_C4S2_M2.csv",
    "../Data/P031/movements_P031_C4S1_M1.csv",
    "../Data/P031/movements_P031_BW_M2.csv",
    # Skip P032
    "../Data/P033/movements_P033_C4S1_M1.csv",
    "../Data/P033/movements_P033_BW_M2.csv",
    "../Data/P034/movements_P034_BW_M1.csv",
    "../Data/P034/movements_P034_C4S1_M2.csv",
    "../Data/P035/movements_P035_C3S2_M2.csv",
    "../Data/P035/movements_P035_BW_M1.csv",
    "../Data/P036/movements_P036_C2S1_M1.csv",
    "../Data/P036/movements_P036_BW_M2.csv",
    "../Data/P037/movements_P037_C4S1_M2.csv",
    "../Data/P037/movements_P037_BW_M1.csv",
    "../Data/P038/movements_P038_C4S2_M2.csv",
    "../Data/P038/movements_P038_BW_M1.csv",
    # Skip P039
    "../Data/P040/movements_P040_C4S1_M2.csv",
    "../Data/P040/movements_P040_BW_M1.csv",
    "../Data/P041/movements_P041_C4S1_M2.csv",
    "../Data/P041/movements_P041_BW_M1.csv",
    "../Data/P042/movements_P042_C4S2_M2.csv",
    "../Data/P042/movements_P042_BW_M1.csv",
    # Skip P043
    "../Data/P044/movements_P044_BW_M1.csv",
    "../Data/P044/movements_P044_C3S3_M2.csv",
    "../Data/P045/movements_P045_BW_M2.csv",
    "../Data/P045/movements_P045_C4S3_M1.csv",
    "../Data/P046/movements_P046_C4S2_M2.csv",
    "../Data/P046/movements_P046_BW_M1.csv",
    "../Data/P047/movements_P047_C4S1_M1.csv",
    "../Data/P047/movements_P047_BW_M2.csv"
]

# Dictionary to store file names and their movement counts
movement_counts = {}
failed_movement_counts = {}

for file in data_files:
    try:
        df = pd.read_csv(file)
        # Number of movements = total rows - 1 (for header)
        num_movements = len(df) - 1
        movement_counts[file] = num_movements
        print(f"{file}: {num_movements} movements")
        tally = 0
        for row in range(len(df)):
            if df['Movement Blocked'][row] == True:
                print(f"  Row {row}: Movement Blocked = TRUE")
                tally+=1
        failed_movement_counts[file] = tally
        print(f"  Failed movements: {failed_movement_counts[file]}")
    except Exception as e:
        print(f"Error reading {file}: {e}")
        movement_counts[file] = None
        failed_movement_counts[file] = None  # Also add None for failed movements

# Print summary
print("\n=== Summary ===")
total_movements = sum([count for count in movement_counts.values() if count is not None])
failed_movements_count = sum([count for count in failed_movement_counts.values() if count is not None])
print(f"Total files processed: {len(movement_counts)}")
print(f"Total movements across all files: {total_movements}")
print(f"Total failed movements: {failed_movements_count}")

# If you want to see the counts in a DataFrame format
print("\n=== Movement Counts ===")
# Create DataFrame properly
summary_data = []
for file in movement_counts.keys():
    summary_data.append({
        'File': file,
        'Movements': movement_counts[file],
        'Failed Movements': failed_movement_counts[file]
    })

summary_df = pd.DataFrame(summary_data)
print(summary_df.to_string())
```

# Export Movement Summary

```python
# Initialize a dictionary to store participant data
participant_data = {}

# Regular expressions to extract participant ID and movement type
pattern_old = r"movements_P(\d{3})_(BW|C\dS\d)_M\d\.csv"
pattern_new = r"movements_P(\d{3})_(BW|C\dS\d)_M\d\.csv"

for file in data_files:
    try:
        df = pd.read_csv(file)
        num_movements = len(df) - 1  # Excluding header
        
        # Try to match with old pattern
        match = re.search(pattern_old, file)
        if not match:
            # Try with new pattern
            match = re.search(pattern_new, file)
        
        if match:
            participant_id = f"P{match.group(1)}"
            movement_type = match.group(2)
            
            # Initialize participant entry if not exists
            if participant_id not in participant_data:
                participant_data[participant_id] = {
                    'Participant_ID': participant_id,
                    'Standard Movements': None,
                    'Personal Movements': None,
                    'Standard Failed Movements': None,
                    'Personal Failed Movements': None
                }
            
            # Assign movements based on type
            if movement_type == 'BW':
                participant_data[participant_id]['Standard Movements'] = num_movements
                participant_data[participant_id]['Standard Failed Movements'] = failed_movement_counts[file]
            elif movement_type.startswith('C'):  # Any CxSx pattern
                participant_data[participant_id]['Personal Movements'] = num_movements
                participant_data[participant_id]['Personal Failed Movements'] = failed_movement_counts[file]
                
        else:
            print(f"Could not parse filename: {file}")
            
    except Exception as e:
        print(f"Error processing {file}: {e}")

# Add participants that were skipped (P001, P016, P023, P032, P039, P043)
skipped_participants = ['P001', 'P016', 'P023', 'P032', 'P039', 'P043']
for pid in skipped_participants:
    if pid not in participant_data:
        participant_data[pid] = {
            'Participant_ID': pid,
            'Standard Movements': None,
            'Personal Movements': None,
            'Standard Failed Movements': None,
            'Personal Failed Movements': None
        }

# Convert to DataFrame and sort by Participant_ID
result_df = pd.DataFrame(list(participant_data.values()))
result_df = result_df.sort_values('Participant_ID').reset_index(drop=True)

# Reorder columns
result_df = result_df[['Participant_ID', 'Standard Movements', 'Personal Movements', 'Standard Failed Movements', 'Personal Failed Movements']]

# Print the results
print("Participant Movement Summary:")
print(result_df.to_string())

# Save to CSV
output_file = 'participant_movements_summary.csv'
result_df.to_csv(output_file, index=False)
print(f"\nResults saved to: {output_file}")
```
